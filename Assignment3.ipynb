3. Write a user-defined function that accepts a list of integers and returns: 
• Count of even numbers 
• Sum of odd numbers 
Use loops and conditional statements inside the function. 

```python
lst= [i for i in range(1,7)]

def check_list(lst):
    even_count = 0
    odd_sum = 0

    for i in lst:
        if i % 2 == 0:
            even_count = even_count + 1
        else:
            odd_sum = odd_sum + i

    return even_count, odd_sum
even,odd=check_list(lst)

print("Even Count =", even)
print("Odd Sum =", odd)




```

4. Create a function that calculates simple interest. 
Use default arguments for rate and time. Demonstrate calling the 
function using both positional and keyword arguments. 

```python
def simple_interest(principal, rate=5, time=2):
    si = (principal * rate * time) / 100
    return si
result1 = simple_interest(1000)
print("Simple Interest (Positional) =", result1)
result2 = simple_interest(principal=2000, rate=10, time=3)
print("Simple Interest (Keyword) =", result2)
```

5. Create a Student class with attributes: name, marks. 
Add a method to calculate grade based on marks. 
Create at least two objects and display their grades. 

```python
class Student:
    def __init__(self, student_name, student_marks):
        self.student_name = student_name
        self.student_marks = student_marks
    def get_grade(self):
        if self.student_marks >= 80:
            grade = "A"
        elif self.student_marks >= 60:
            grade = "B"
        elif self.student_marks >= 40:
            grade = "C"
        else:
            grade = "Fail"
        return grade

s1 = Student("Aman", 92)
s2 = Student("Neha", 58)
print("Student-1 Details:-")
print("Name:-",s1.student_name)
print("Marks:-",s1.student_marks)
print("Grade:-",s1.get_grade())
print()
print("Student-2 Details:-")
print("Name:-",s2.student_name)
print("Marks:-",s2.student_marks)
print("Grade:-",s2.get_grade())
```

6. Create a BankAccount class with private attributes for balance. 
Provide methods to deposit and withdraw money. Ensure balance 
cannot be accessed directly. 

```python
class BankAccount:

    def __init__(self, initial_balance):
        self.__balance = initial_balance   
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount Deposited:", amount)
        else:
            print("Invalid Deposit Amount")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Amount Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def show_balance(self):
        print("Current Balance:", self.__balance)

account1 = BankAccount(1000)
account1.deposit(500)
account1.withdraw(300)
account1.show_balance()

```

8. Create a custom exception that raises an error if a user enters age less 
than 0 or greater than 120. 

```python

class InvalidAgeError(Exception):
    pass

try:
    age = int(input("Enter your age: "))

  
    if age < 0 or age > 120:
        raise InvalidAgeError("Age must be between 0 and 120.")

    print("Valid age entered:", age)


except InvalidAgeError as e:
    print("Error:", e)

except ValueError:
    print("Error: Please enter a valid number.")

finally:
    print("Program ended.")
```

10. Write a program that reads numbers from a file and calculates: 
• Total 
• Average 
Handle the case if the file does not exist. 

```python
try:
    file_path = r"D:\Data Science Tranning\Assignment3\numbers.txt"

    total = 0
    count = 0

    with open(file_path, "r") as file:
        for line in file:
            numbers = line.split()
            for num in numbers:
                total += float(num)
                count += 1

    average = total / count if count > 0 else 0

    print("Total:", total)
    print("Average:", average)

except FileNotFoundError:
    print("Error: File does not exist.")

except ValueError:
    print("Error: File contains invalid data.")
```
